```python
import pandas as pd
import math
```

```python
data = {'V1':     [0.1, 0.2,  0.18, 0.28],
        'V2':     [0.3, 0.01, 0.1,  0.32],
        'classe': ['A',  'B', 'B',   'A']}
df = pd.DataFrame (data, columns = ['V1','V2', 'classe'])
```

```python
df
```

```python
novo_data = {'V1':     [0.15],
        'V2':     [0.31],
        'classe': ['?']}
novo = pd.DataFrame(novo_data)
print(novo)
```

```python
d0 = math.sqrt((novo['V1'][0] - df['V1'][0])**2 + (novo['V2'][0] - df['V2'][0])**2)
d1 = math.sqrt((novo['V1'][0] - df['V1'][1])**2 + (novo['V2'][0] - df['V2'][1])**2)
d2 = math.sqrt((novo['V1'][0] - df['V1'][2])**2 + (novo['V2'][0] - df['V2'][2])**2)
d3 = math.sqrt((novo['V1'][0] - df['V1'][3])**2 + (novo['V2'][0] - df['V2'][3])**2)
print("D0: ", d0) 
print("D1: ", d1)
print("D2: ", d2)
print("D3: ", d3)
```
